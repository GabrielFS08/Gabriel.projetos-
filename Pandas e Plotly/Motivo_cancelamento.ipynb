```python
#importar a base de dados:
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")

#remover o CostumerID por ser irrelevante:
tabela = tabela.drop(columns="CustomerID")
display(tabela)
```

```python
#mostra as informações da tabela:
display(tabela.info())
```

```python
#remover as informações vazias:
tabela = tabela.dropna()
display(tabela.info())
```

```python
#mostra pessoas que cancelaram e não cancelaram em quantidade:
display(tabela["cancelou"].value_counts())

#mostra pessoas que cancelaram e não cancelaram em porcentagem:
#normalize=True faz mostrar em porcentagem, e o .map("{:.1%}".format) faz mostrar somente uma casa decimal depois da vírgula
display(tabela["cancelou"].value_counts(normalize=True).map("{:.1%}".format))
```

```python
#mostra em porcentagem as pessoas que estão em cada tipo de contrato:
display(tabela["duracao_contrato"].value_counts(normalize=True).map("{:.1%}".format))
display(tabela["duracao_contrato"].value_counts())
```

```python
#agrupa as informações da coluna e faz uma média das informações de colunas da tabela:
display(tabela.groupby("duracao_contrato").mean(numeric_only=True))
```

- TODOS QUE SÃO DO CONTRATO MENSAL CANCELARAM O SERVIÇO!!

```python
#removeremos o contrato Monthly	para saber o impacto que ele vai fazer na coluna de cancelou:
tabela = tabela[tabela["duracao_contrato"] != "Monthly"]
display(tabela)
#agora não temos o valor Monthly na coluna de duracao_contrato
```

```python
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:.1%}".format))
#observamos o impacto que remover o Monthly fez na coluna Cancelou:
```

- DE 56% CAIU PRA 46%

```python
#agora vamos observar como a assinatura está distribuída na tabela em porcentagem:
display(tabela["assinatura"].value_counts(normalize=True).map("{:.1%}".format))
#e a sua média em comparação é:
display(tabela.groupby("assinatura").mean(numeric_only=True))
```

- AS PROPORÇÕES SÃO QUASE AS MESMAS, NÃO DEVEMOS JULGA LAS PELA ASSINATURA.

```python
#criaremos um gráfico para melhor visualização:
import plotly.express as px

for coluna in tabela: #o padrão itera sobre colunas da base de dados
    grafico = px.histogram(tabela, x=coluna, color="cancelou", width=600)
    grafico.show()
```

- DEPOIS DE 20 DIAS DE ATRASO O CLIENTE CANCELA.
- DEPOIS DE 5 LIGAÇÕES O CLIENTE CANCELA.
- PESSOAS ACIMA DE 50 ANOS CANCELAM.

```python
#iremos remover essas informações para observar a diferença:
tabela = tabela[tabela["ligacoes_callcenter"]<5]#ira contar apenas os clientes com ligações menores que 5...
tabela = tabela[tabela["dias_atraso"]<=20]
tabela = tabela[tabela["idade"]<50]
display(tabela)
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:.1%}".format))
```
