```python
import pandas as pd
tabela = pd.read_csv("clientes.csv") #importando a base de dados
display(tabela)
```

```python
#verfificar se tem valores vazios ou valores errados
print(tabela.info())
print(tabela.columns)
```

```python
from sklearn.preprocessing import LabelEncoder

#vai transformar os textos em números da coluna específica
codificador_profissao = LabelEncoder()
tabela["profissao"] = codificador_profissao.fit_transform(tabela["profissao"])#o novo valor da coluna "profissao" é igual ao seu valor transformado em int com a ajuda do aplicador ".fit_transform"

codificador_mix_credito = LabelEncoder()
tabela["mix_credito"] = codificador_mix_credito.fit_transform(tabela["mix_credito"])

codificador_comportamento = LabelEncoder()
tabela["comportamento_pagamento"] = codificador_comportamento.fit_transform(tabela["comportamento_pagamento"])
```

```python
#vamos escolher quais colunas vamos usar para treinar o modelo
#y é a coluna que vamos prever(a coluna"score_credito")
#x é a coluna que vamos usar para prever o score de crédito, não vamos usar a coluna id_cliente porque ela é um número qualquer e não ajuda na previsão
y = tabela["score_credito"]
x = tabela.drop(["score_credito", "id_cliente"], axis=1)#estou excluindo essas duas colunas usando o .drop(), e o axis=1 significa que eu estou escolhendo a coluna, e não a linha(linha: axis = 0).

from sklearn.model_selection import train_test_split

#separamos os dados em treino e teste, treino vamos dar para os modelos aprederem e teste vamos usar para ver se o modelo aprendeu corretamente
x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.3, random_state=1)
#test_size=0.3 siginifca que 30% dos dados oferecidos seram usados como teste.
#e o random_state=1, sigifica que sempre que eu rodar esse código o resultado vai ser aleatório.
```

```python
#modelo de árvore de decisão
from sklearn.ensemble import RandomForestClassifier
#modelo de vizinhos mais próximos
from sklearn.neighbors import KNeighborsClassifier

modelo_arvore = RandomForestClassifier()
modelo_knn = KNeighborsClassifier()

#treinando os modelos
modelo_arvore.fit(x_treino, y_treino)
modelo_knn.fit(x_treino, y_treino)

#se ele chutasse tudo "Standard", qual seria seu nível de acerto?
contagem_scores = tabela["score_credito"].value_counts()
print(contagem_scores["Standard"] / sum(contagem_scores))
```

```python
from sklearn.metrics import accuracy_score

#testamos os dois modelos
previsao_arvore = modelo_arvore.predict(x_teste)
previsao_knn = modelo_knn.predict(x_teste)

#comparamos o resultado com  o y_teste
display(accuracy_score(y_teste, previsao_arvore))
display(accuracy_score(y_teste, previsao_knn))

#como melhorar o modelo:
#GridSearch
#ter mais dados
#treinar outros parâmetros
```

```python
#quais as características mais importantes para definir o score de crédito?
coluna = list(x_teste.columns)
importancia = pd.DataFrame(index=coluna, data=modelo_arvore.feature_importances_)
importancia = importancia * 100
print(importancia)
```

```python
#agora vamos testar o modelo de árvore de decisão nesse novos clientes
#importar os novos clientes
novos_clientes = pd.read_csv("novos_clientes.csv")

#transformar os dados em str para números inteiros
#agora é so "transform", porque ja tem rotulos de int criados anteriormente e também porque iria sobreescrever os int ja criados
novos_clientes["profissao"] = codificador_profissao.transform(novos_clientes["profissao"])

novos_clientes["mix_credito"] = codificador_mix_credito.transform(novos_clientes["mix_credito"])

novos_clientes["comportamento_pagamento"] = codificador_comportamento.transform(novos_clientes["comportamento_pagamento"])

#fazer a previsão com o modelo escolhido
previsao = modelo_arvore.predict(novos_clientes)
display(novos_clientes)
display(previsao)
```
