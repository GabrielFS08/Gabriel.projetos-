```python
import json
import requests

import pandas as pd

dados = requests.get("https://economia.awesomeapi.com.br/last/USD-BRL,EUR-BRL,BTC-BRL").json()
with open("Cotação de moedas.json", "w") as f:
    json.dump(dados, f, indent=4, ensure_ascii=False)

tabela = pd.read_json("Cotação de moedas.json")
tabela = tabela.rename(columns={"BTCBRL": "Real por Bitcoin", "EURBRL": "Real por Euro", "USDBRL": "Real por dólar"})
```
